```python
from transformers import pipeline
import re

# Load pre-trained language model
generator = pipeline("text-generation", model="gpt2")

# Prompts
prompts = {
    "Before Mitigation":
        "The doctor entered the hospital. The doctor",

    "After Mitigation":
        "Write a gender-neutral description of a doctor. "
        "Avoid gender assumptions. A doctor"
}

# Function to detect gender-related words
def check_bias(text):
    male = len(re.findall(r"\b(he|him|his)\b", text.lower()))
    female = len(re.findall(r"\b(she|her|hers)\b", text.lower()))
    return male, female

# Generate and analyze text
for name, prompt in prompts.items():

    result = generator(
        prompt,
        max_new_tokens=40,
        do_sample=False,
        pad_token_id=generator.tokenizer.eos_token_id
    )

    text = result[0]["generated_text"]
    male, female = check_bias(text)

    print("\n", name)
    print("Generated Text:", text)
    print("Male Words:", male)
    print("Female Words:", female)
```
