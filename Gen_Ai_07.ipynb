```python
import tensorflow as tf
from tensorflow.keras import layers, Model

# define parameters

seq_len = 5
embed_dim = 8
num_heads = 2

inputs = layers.Input(shape=(seq_len,embed_dim))

attention = layers.MultiHeadAttention(
    num_heads = num_heads,key_dim=embed_dim//num_heads)(inputs,inputs)


x=layers.LayerNormalization()(inputs+attention)
ffn = layers.Dense(16,activation="relu")(x)
ffn = layers.Dense(embed_dim)(ffn)\


outputs = layers.LayerNormalization()(x+ffn)


model = Model(inputs,outputs)
sample = tf.random.normal((1,seq_len,embed_dim))

result = model(sample)

print("Input Shape :",sample.shape)
print("Output Shape :",result.shape)
print("Number of Attention Heads :",num_heads)
```
